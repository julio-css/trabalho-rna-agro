## 2. Estrutura do Neurônio Biológico

```python
# Desenhar neurônio biológico
fig, ax = plt.subplots(figsize=(12, 8))
ax.set_xlim(-2, 12)
ax.set_ylim(-2, 10)
ax.axis('off')

# Dendrites (entrada)
for i in range(3):
    x_start = 0
    y_start = 6 + i*1.5
    # Ramificações
    for j in range(4):
        x_branch = x_start - 1 - j*0.5
        y_branch = y_start + np.random.randn()*0.3
        ax.plot([x_start, x_branch], [y_start, y_branch], 'b-', linewidth=2, alpha=0.7)
    # Ramo principal
    ax.plot([x_start, 1.5], [y_start, 5], 'b-', linewidth=3)
    ax.text(-1.5, y_start, f'Input {i+1}', fontsize=10, color='blue', fontweight='bold')

# Soma (corpo celular)
soma = Circle((2.5, 5), 0.6, color='gold', ec='black', linewidth=2.5, zorder=10)
ax.add_patch(soma)
ax.text(2.5, 5, 'Soma\n(Núcleo)', fontsize=10, ha='center', va='center', fontweight='bold')

# Axon hillock
ax.plot([3.1, 3.5], [5, 5], 'k-', linewidth=3)
ax.text(3.3, 5.4, 'Axon\nHillock', fontsize=8, ha='center')

# Axon (saída)
for i in range(6):
    x = 3.5 + i*1.2
    y = 5 + 0.3*np.sin(i*0.5)
    if i == 0:
        ax.plot([3.5, x], [5, y], 'r-', linewidth=3)
    else:
        ax.plot([3.5 + (i-1)*1.2, x], [5 + 0.3*np.sin((i-1)*0.5), y], 'r-', linewidth=3)

# Axon terminals (sinapses)
for i in range(3):
    x_terminal = 9
    y_terminal = 7 - i*1.5
    ax.plot([7.5, x_terminal], [5, y_terminal], 'r-', linewidth=2)
    # Vesículas (transmissores)
    for j in range(2):
        circ = Circle((x_terminal - 0.3 + j*0.2, y_terminal + 0.1), 0.1, 
                     color='purple', alpha=0.7)
        ax.add_patch(circ)
    ax.text(x_terminal + 0.5, y_terminal, f'Output {i+1}', fontsize=10, 
           color='red', fontweight='bold')

# Labels
ax.text(2.5, 3.5, 'Dendrites\n(Recebem)', fontsize=10, ha='center', 
       bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.7))
ax.text(7, 3.5, 'Axon\n(Envia)', fontsize=10, ha='center',
       bbox=dict(boxstyle='round', facecolor='lightcoral', alpha=0.7))

ax.set_title('Estrutura do Neurônio Biológico', fontsize=14, fontweight='bold', pad=20)
plt.tight_layout()
plt.savefig('results/plots/neuron_biological.png', dpi=150, bbox_inches='tight')
plt.show()

print("\n📐 PARTES DO NEURÔNIO:")
print("\n1. DENDRITES (Entrada):")
print("   • Recebem sinais de outros neurônios")
print("   • Múltiplas ramificações (aumenta área de recepção)")
print("   • Receptores especializados para diferentes químicos")

print("\n2. SOMA (Corpo):")
print("   • Contém o núcleo (DNA)")
print("   • Integra sinais chegando pelos dendrites")
print("   • Decide se vai 'dispara'")

print("\n3. AXON HILLOCK:")
print("   • Zona de disparo")
print("   • Se voltagem atinge limiar → dispara potencial de ação")

print("\n4. AXON (Saída):")
print("   • Transmite sinal para longe do soma")
print("   • Pode ter até 1 metro de comprimento")
print("   • Ramifica no final (axon terminals)")

print("\n5. SINAPSES (Conexões):")
print("   • Espaço entre axon terminal e dendrite")
print("   • Transmissão química (neurotransmissores)")
print("   • Podem ser excitatórias (aumentam) ou inibitórias (diminuem)")
```

## 3. Potencial de Ação: Como Neurônios "Disparam"

```python
# Simular potencial de ação
time = np.linspace(0, 5, 1000)

# Fases do potencial de ação
voltage = np.zeros_like(time)
for i, t in enumerate(time):
    if t < 1:
        voltage[i] = -70 + 30*t  # Depolarização
    elif t < 1.5:
        voltage[i] = -40 - 120*(t-1)  # Repolarização
    elif t < 2:
        voltage[i] = -70 - 30*(t-1.5)  # Hiperpolarização
    else:
        voltage[i] = -70  # Repouso

# Plotar
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))

# Voltagem ao longo do tempo
ax1.plot(time, voltage, 'b-', linewidth=2.5)
ax1.axhline(y=-70, color='k', linestyle='--', alpha=0.5, label='Repouso (-70mV)')
ax1.axhline(y=-55, color='r', linestyle='--', alpha=0.5, label='Limiar (-55mV)')
ax1.fill_between(time, voltage, -70, where=(voltage >= -70), alpha=0.2, color='red', label='Potencial de Ação')
ax1.set_ylabel('Voltagem (mV)', fontweight='bold', fontsize=11)
ax1.set_title('Potencial de Ação: Como Neurônios Disparam', fontweight='bold', fontsize=12)
ax1.legend(loc='upper left')
ax1.grid(True, alpha=0.3)
ax1.set_xlim(0, 5)

# Fases
fases = [
    (0.5, -50, 'Repouso\n(-70mV)', 'gray'),
    (1.2, 20, 'Depolarização\n(+30mV)', 'red'),
    (1.75, -90, 'Repolarização\n(-120mV)', 'orange'),
    (2.25, -100, 'Hiperpolarização\n(-100mV)', 'purple'),
    (3.5, -70, 'Retorno\n(-70mV)', 'green')
]

for x, y, label, color in fases:
    ax1.scatter(x, y, s=150, color=color, edgecolor='black', linewidth=2, zorder=5)
    ax1.annotate(label, (x, y), xytext=(x, y+25), fontsize=9, ha='center',
                arrowprops=dict(arrowstyle='->', color=color, lw=1.5))

# Timeline com eventos
ax2.set_ylim(0, 4)
ax2.set_xlim(0, 5)
ax2.axis('off')

events = [
    (0.5, 'Estímulo chega\npelos dendrites'),
    (1, 'Soma integra sinais'),
    (1.2, '✓ Atinge limiar\nDispara!'),
    (1.75, 'Potencial de ação\npropaga pelo axon'),
    (2.5, 'Neurotransmissores\nliberados na sinapse'),
    (3.5, 'Período refratário\n(não pode disparar)')
]

for x, label in events:
    ax2.scatter(x, 2, s=200, color='blue', zorder=5)
    ax2.plot([x, x], [1.5, 2.5], 'k-', linewidth=1)
    ax2.text(x, 3, label, fontsize=9, ha='center',
            bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))

ax2.set_title('Timeline: Vida de um Potencial de Ação', fontweight='bold', fontsize=12)
ax2.set_xlabel('Tempo (ms)', fontweight='bold')

plt.tight_layout()
plt.savefig('results/plots/action_potential.png', dpi=150, bbox_inches='tight')
plt.show()

print("\n⚡ POTENCIAL DE AÇÃO: Sequência de Eventos")
print("\n1. REPOUSO (-70mV):")
print("   • Voltagem em repouso do neurônio")
print("   • Pronto para receber estímulos")

print("\n2. ESTIMULAÇÃO:")
print("   • Sinais chegam pelos dendrites")
print("   • Se soma recebe MUITOS sinais excitatórios...")

print("\n3. DESPOLARIZAÇÃO (sobe para +30mV):")
print("   • Voltagem aumenta rapidamente")
print("   • Canais de Na+ se abrem (entra Na+)")
print("   • Se atinge limiar (-55mV) → DISPARA!")

print("\n4. REPOLARIZAÇÃO (desce para -90mV):")
print("   • Canais de Na+ fecham, K+ abre")
print("   • Voltagem volta rapidamente")

print("\n5. HIPERPOLARIZAÇÃO:")
print("   • Vai abaixo do repouso (-100mV)")
print("   • Retorna gradualmente")

print("\n6. PERÍODO REFRATÁRIO:")
print("   • Neurônio não pode disparar novamente")
print("   • Evita feedback infinito")
print("   • Duração: ~1-2 ms")

print("\n⏱️  VELOCIDADE:")
print("   • Potencial de ação: ~1ms (muito rápido)")
print("   • Transmissão sináptica: ~1-5ms")
print("   • Total: ~2-7ms entre neurônios")
```

## 4. Sinapses: Como Neurônios Comunicam

```python
# Desenhar sinapse em detalhes
fig, ax = plt.subplots(figsize=(12, 7))
ax.set_xlim(0, 10)
ax.set_ylim(0, 8)
ax.axis('off')

# Neurônio pré-sináptico (enviador)
ax.text(1, 7.5, 'Neurônio Pré-Sináptico\n(Enviador)', fontsize=11, fontweight='bold',
       ha='center', bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.8))
ax.add_patch(Circle((1, 6), 0.4, color='lightblue', ec='black', linewidth=2))
ax.plot([1, 1], [5.6, 4.5], 'b-', linewidth=3)

# Axon terminal
ax.add_patch(Circle((1, 4.2), 0.3, color='darkblue', ec='black', linewidth=2))
ax.text(1.8, 4.2, 'Axon Terminal', fontsize=9, fontweight='bold')

# Vesículas com neurotransmissores
for i in range(5):
    x = 0.7 + i*0.15
    y = 4.5 + np.random.randn()*0.1
    ax.add_patch(Circle((x, y), 0.08, color='purple', alpha=0.8))
ax.text(1.5, 5, 'Vesículas\n(Neurotransmissores)', fontsize=9, fontweight='bold')

# Fenda sináptica
rect = Rectangle((0.5, 3.5), 1, 0.6, color='lightgreen', alpha=0.5, edgecolor='green', linewidth=2)
ax.add_patch(rect)
ax.text(1, 3.2, 'Fenda Sináptica\n(20-40 nm)', fontsize=9, ha='center', fontweight='bold')

# Receptores
for i in range(5):
    x = 2 + i*0.15
    y = 3.5 + np.random.randn()*0.1
    ax.scatter(x, y, s=100, marker='s', color='red', edgecolor='darkred', linewidth=1.5)
ax.text(3, 4.2, 'Receptores\n(Proteínas)', fontsize=9, fontweight='bold')

# Neurônio pós-sináptico (receptor)
ax.plot([2, 2], [2.9, 2.2], 'r-', linewidth=3)
ax.add_patch(Circle((2, 2), 0.4, color='lightcoral', ec='black', linewidth=2))
ax.text(2, 1, 'Neurônio Pós-Sináptico\n(Receptor)', fontsize=11, fontweight='bold',
       ha='center', bbox=dict(boxstyle='round', facecolor='lightcoral', alpha=0.8))

# Setas de processo
arrow1 = FancyArrowPatch((1.2, 4.5), (1.5, 4), arrowstyle='->', 
                        mutation_scale=20, linewidth=2, color='purple')
ax.add_patch(arrow1)
ax.text(1.8, 4.4, '1. Liberação', fontsize=8, style='italic')

arrow2 = FancyArrowPatch((1.5, 3.2), (1.8, 3.2), arrowstyle='->', 
                        mutation_scale=20, linewidth=2, color='green')
ax.add_patch(arrow2)
ax.text(1.65, 2.9, '2. Difusão', fontsize=8, style='italic')

arrow3 = FancyArrowPatch((2.2, 2.8), (2, 2.4), arrowstyle='->', 
                        mutation_scale=20, linewidth=2, color='red')
ax.add_patch(arrow3)
ax.text(2.4, 2.6, '3. Ligação', fontsize=8, style='italic')

ax.set_title('Sinapse Química: Transmissão Entre Neurônios', fontsize=13, fontweight='bold', pad=20)
plt.tight_layout()
plt.savefig('results/plots/synapse_detailed.png', dpi=150, bbox_inches='tight')
plt.show()

print("\n🔗 TRANSMISSÃO SINÁPTICA: 3 Passos")
print("\n1. LIBERAÇÃO (Pré-sináptico):")
print("   • Potencial de ação chega ao axon terminal")
print("   • Ca²⁺ entra no terminal")
print("   • Vesículas se fundem com membrana")
print("   • Neurotransmissores liberados na fenda (~1000 moléculas)")
print("   • Tempo: ~0.5 ms")

print("\n2. DIFUSÃO (Fenda sináptica):")
print("   • Neurotransmissores atravessam fenda (20-40 nanômetros)")
print("   • Movimento aleatório (difusão)")
print("   • Tempo: ~1 ms")

print("\n3. RECEPÇÃO (Pós-sináptico):")
print("   • Neurotransmissor se liga a receptor")
print("   • Receptor é uma proteína no dendrite")
print("   • Abre canais de íons")
print("   • Gera potencial pós-sináptico")
print("   • Tempo: ~1 ms")

print("\n📊 TIPOS DE SINAPSES:")
print("   • Excitatória: aumenta probabilidade de disparo (AMPA, NMDA)")
print("   • Inibitória: diminui probabilidade de disparo (GABA)")
print("   • Razão típica: 4 excitatórias : 1 inibitória")

print("\n⚖️ INTEGRAÇÃO:")
print("   • Soma recebe 7.000+ sinais")
print("   • Integra tudo (soma temporal e espacial)")
print("   • Se soma dos sinais > limiar → DISPARA")
print("   • Caso contrário → silencia")
```

## 5. Plasticidade Sináptica: Como o Cérebro Aprende

```python
print("\n" + "="*70)
print("🧠 PLASTICIDADE SINÁPTICA: O Cérebro Aprende Mudando")
print("="*70)

plasticity_types = {
    "Long-Term Potentiation (LTP)": {
        "o_que_e": "Fortalecimento de sinapses",
        "como_funciona": "Uso repetido aumenta a força da conexão",
        "timescale": "Minutos a horas",
        "mecanismo": "↑ Receptores de AMPA, ↑ Neurotransmissores",
        "implicacao": "Memória de curto prazo → longo prazo",
        "exemplo": "Aprender a dirigir: primeiras vezes difícil, depois automático"
    },
    "Long-Term Depression (LTD)": {
        "o_que_e": "Enfraquecimento de sinapses",
        "como_funciona": "Falta de uso enfraquece a conexão",
        "timescale": "Minutos a horas",
        "mecanismo": "↓ Receptores, ↓ Libração de neurotransmissores",
        "implicacao": "Esquecer informações não importantes",
        "exemplo": "Não praticar piano → piora a performance"
    },
    "Spike-Timing-Dependent Plasticity (STDP)": {
        "o_que_e": "Aprendizado baseado em timing",
        "como_funciona": "Se pré-sináptico dispara ANTES de pós: LTP\nSe DEPOIS: LTD",
        "timescale": "Milissegundos",
        "mecanismo": "Correlação temporal entre neurônios",
        "implicacao": "Aprendizado por associação",
        "exemplo": "Associar luz com som (condicionamento clássico)"
    }
}

for tipo, detalhes in plasticity_types.items():
    print(f"\n{tipo.upper()}:")
    for chave, valor in detalhes.items():
        print(f"  • {chave}: {valor}")

print("\n" + "="*70)
print("🔑 REGRA DE HEBB (1949): 'Neurons that fire together, wire together'")
print("="*70)
print("""
Impactante descoberta:
  • Se neurônio A e B disparam juntos repetidamente
  • A sinapse entre eles se fortalece
  • Próxima vez que A dispara, mais provável B disparar também
  
Exemplo real:
  • Ouve campainha + comida → cão associa
  • Campainha sozinha → salivação (condicionado)
  • Sinapses entre neurônios auditivos e gustativos se fortaleceram

Implicação para RNA:
  • Backpropagation é inspirado em STDP!
  • Ajuste de pesos = ajuste de força de sinapses
  • Dados correlacionados = padrões aprendidos
""")

# Visualizar plasticidade
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

# LTP
ax = axes[0]
time = np.linspace(0, 10, 100)
force = np.ones_like(time)
for i, t in enumerate(time):
    force[i] = 1 + 0.8 * (1 - np.exp(-t/3))  # Crescimento exponencial

ax.plot(time, force, 'g-', linewidth=3)
ax.fill_between(time, 1, force, alpha=0.3, color='green')
ax.axhline(y=1, color='k', linestyle='--', alpha=0.5)
ax.set_title('LTP: Fortalecimento', fontweight='bold')
ax.set_xlabel('Uso repetido')
ax.set_ylabel('Força da sinapse')
ax.grid(True, alpha=0.3)
ax.text(5, 1.3, 'Memória\nforma-se', fontsize=10, ha='center',
       bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.8))

# LTD
ax = axes[1]
force = np.ones_like(time)
for i, t in enumerate(time):
    force[i] = 1 - 0.6 * (1 - np.exp(-t/3))

ax.plot(time, force, 'r-', linewidth=3)
ax.fill_between(time, force, 1, alpha=0.3, color='red')
ax.axhline(y=1, color='k', linestyle='--', alpha=0.5)
ax.set_title('LTD: Enfraquecimento', fontweight='bold')
ax.set_xlabel('Falta de uso')
ax.set_ylabel('Força da sinapse')
ax.grid(True, alpha=0.3)
ax.text(5, 0.5, 'Memória\napaga-se', fontsize=10, ha='center',
       bbox=dict(boxstyle='round', facecolor='lightcoral', alpha=0.8))

# STDP
ax = axes[2]
timing = np.linspace(-50, 50, 100)  # ms
change = 100 * np.exp(-np.abs(timing)/20) * np.sign(timing)  # Assimétrico

ax.plot(timing, change, 'b-', linewidth=3)
ax.fill_between(timing, 0, change, where=(change>=0), alpha=0.3, color='green', label='LTP')
ax.fill_between(timing, 0, change, where=(change<0), alpha=0.3, color='red', label='LTD')
ax.axhline(y=0, color='k', linestyle='-', linewidth=1)
ax.axvline(x=0, color='k', linestyle='--', alpha=0.5)
ax.set_title('STDP: Timing é tudo', fontweight='bold')
ax.set_xlabel('Timing pré - pós (ms)')
ax.set_ylabel('Mudança de força')
ax.grid(True, alpha=0.3)
ax.legend()

plt.suptitle('Plasticidade Sináptica: 3 Mecanismos de Aprendizado', 
            fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('results/plots/synaptic_plasticity.png', dpi=150, bbox_inches='tight')
plt.show()
```
