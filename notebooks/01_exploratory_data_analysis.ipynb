## 1. Explorar PlantDoc (Milho + Soja)

```python
from PIL import Image

plantdoc_path = Path("data/raw/PlantDoc-Dataset/raw/color")

if plantdoc_path.exists():
    print("PlantDoc encontrado!")
    print()
    
    # Contar imagens por classe
    class_counts = {}
    for class_dir in plantdoc_path.iterdir():
        if class_dir.is_dir():
            count = len(list(class_dir.glob("*.jpg")))
            class_counts[class_dir.name] = count
    
    # Mostrar estatísticas
    df_plantdoc = pd.DataFrame(list(class_counts.items()), columns=['Classe', 'Imagens'])
    print("Distribuição de imagens por classe:")
    print(df_plantdoc.to_string(index=False))
    print(f"\nTotal: {df_plantdoc['Imagens'].sum()} imagens")
else:
    print(f"❌ PlantDoc não encontrado em {plantdoc_path}")
```

```python
# Visualizar algumas imagens de PlantDoc
if plantdoc_path.exists():
    fig, axes = plt.subplots(2, 3, figsize=(12, 8))
    axes = axes.flatten()
    
    img_count = 0
    for class_dir in sorted(plantdoc_path.iterdir()):
        if class_dir.is_dir() and img_count < 6:
            img_path = list(class_dir.glob("*.jpg"))[0]
            img = Image.open(img_path)
            
            ax = axes[img_count]
            ax.imshow(img)
            ax.set_title(class_dir.name, fontsize=10, fontweight='bold')
            ax.axis('off')
            
            img_count += 1
    
    plt.suptitle('PlantDoc - Exemplos de classes', fontsize=14, fontweight='bold')
    plt.tight_layout()
    plt.savefig('results/plots/01_plantdoc_samples.png', dpi=150, bbox_inches='tight')
    plt.show()
```

## 2. Explorar BRACOL (Café Arábica)

```python
bracol_path = Path("data/raw/bracol")

if bracol_path.exists():
    print("BRACOL encontrado!")
    print("Estrutura de diretórios:")
    for item in bracol_path.rglob("*"):
        if item.is_file() and item.suffix.lower() in ['.jpg', '.png']:
            print(f"  {item.relative_to(bracol_path)}")
            break
else:
    print(f"❌ BRACOL não encontrado em {bracol_path}")
```
