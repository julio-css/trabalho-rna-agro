## Setup

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.patches import Circle, Rectangle, FancyArrowPatch, FancyBboxPatch

plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

print("✓ Setup concluído")
```

## 1. Neurônio Biológico Simplificado

```python
print("\n" + "="*70)
print("🧠 NEURÔNIO BIOLÓGICO: O PROCESSO")
print("="*70)

print("""
1. RECEPÇÃO (Dendrites):
   • Recebe sinais de múltiplos neurônios
   • Cada sinapse tem "força" diferente
   • Sinais excitatórios (+) vs inibitórios (-)

2. INTEGRAÇÃO (Soma):
   • Soma todos os sinais chegando
   • Se TOTAL > limiar → dispara
   • Fórmula: V_soma = Σ(força_sinapse × sinal_entrada)

3. TRANSMISSÃO (Axon):
   • Se disparou → envia potencial de ação
   • Propaga para axon terminals

4. TRANSMISSÃO QUÍMICA (Sinapses):
   • Libera neurotransmissores
   • Alcançam outros neurônios (pós-sinápticos)
   • Ciclo se repete

EQUAÇÃO SIMPLIFICADA DO NEURÔNIO BIOLÓGICO:

  V_entrada = w₁×sinal₁ + w₂×sinal₂ + ... + wₙ×sinalₙ + bias
  
  Onde:
    • w_i = força da sinapse (plasticidade!)
    • sinal_i = input do neurônio pré-sináptico
    • bias = "facilidade" de disparar
    
  Se V_entrada > limiar → dispara (output = 1)
  Caso contrário → silencia (output = 0)
""")

print("="*70)
```

## 2. Neurônio Artificial

```python
print("\n" + "="*70)
print("🤖 NEURÔNIO ARTIFICIAL: O PROCESSO")
print("="*70)

print("""
1. MULTIPLICAÇÃO (Pesos):
   • Cada entrada x_i é multiplicada por peso w_i
   • Peso = força da "sinapse artificial"
   • Pesos são números aprendidos

2. SOMA:
   • Soma todos: z = Σ(w_i × x_i) + b
   • b = bias (como limiar biológico)

3. ATIVAÇÃO (Função não-linear):
   • Aplica função σ(z) [sigma]
   • output = σ(z) = σ(Σ(w_i × x_i) + b)
   • Comum: ReLU(z) = max(0, z) ou Sigmoid

4. PROPAGAÇÃO (Para próxima camada):
   • Output de um neurônio = input do próximo
   • Multiplica pelos pesos da próxima camada
   • Ciclo se repete

EQUAÇÃO DO NEURÔNIO ARTIFICIAL:

  z = w₁×x₁ + w₂×x₂ + ... + wₙ×xₙ + b    (linear)
  ŷ = σ(z)                                 (não-linear)
  
  Onde:
    • x_i = entrada
    • w_i = peso (aprendido por backprop)
    • b = bias
    • σ = função de ativação (ReLU, Sigmoid, Tanh)
    • ŷ = output
""")

print("="*70)
```

## 3. Comparação Lado a Lado

```python
# Tabela de comparação
comparison = {
    "Aspecto": [
        "Entrada",
        "Integração",
        "Força da conexão",
        "Fórmula",
        "Saída",
        "Aprendizado",
        "Velocidade",
        "Robustez",
        "Paralelismo"
    ],
    "Neurônio Biológico": [
        "Neurotransmissores (químico)",
        "Soma de sinais em potencial de membrana",
        "Força sináptica (plasticidade)",
        "V_soma = Σ(w×sinal) + bias",
        "Tudo-ou-nada (dispara ou não)",
        "LTP/LTD (plasticidade sináptica)",
        "~200 Hz",
        "Muito robusto (ruído natural)",
        "Massivo (86B neurônios em paralelo)"
    ],
    "Neurônio Artificial": [
        "Números (elétrico, no computador)",
        "Produto escalar (operação matemática)",
        "Peso (w) do modelo",
        "z = Σ(w×x) + b; ŷ = σ(z)",
        "Contínuo (0-1 com Sigmoid, etc)",
        "Backpropagation (gradient descent)",
        "~1 Petaflop/s (computador)",
        "Frágil (precisa dados bons)",
        "Maciço em GPU (bilhões de ops/s)"
    ]
}

print("\n" + "="*70)
print("📊 COMPARAÇÃO: NEURÔNIO BIOLÓGICO vs ARTIFICIAL")
print("="*70)

for i, aspecto in enumerate(comparison["Aspecto"]):
    print(f"\n{aspecto.upper()}:")
    print(f"  🧠 Bio:   {comparison['Neurônio Biológico'][i]}")
    print(f"  🤖 IA:    {comparison['Neurônio Artificial'][i]}")

print("\n" + "="*70)
```

## 4. Visualização: Bio vs Artificial

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# NEURÔNIO BIOLÓGICO
ax1.set_xlim(-1, 8)
ax1.set_ylim(-1, 8)
ax1.axis('off')
ax1.set_title('Neurônio Biológico', fontsize=12, fontweight='bold', pad=10)

# Entradas
for i in range(3):
    y = 6 - i*1.5
    ax1.plot([0, 1], [y, 5], 'b-', linewidth=2)
    ax1.scatter(0, y, s=80, color='blue', zorder=5)
    ax1.text(-0.5, y, f'Input {i+1}', fontsize=9)

# Soma
ax1.add_patch(Circle((2, 5), 0.4, color='gold', ec='black', linewidth=2))
ax1.text(2, 5, 'Σ', fontsize=12, ha='center', va='center', fontweight='bold')
ax1.text(2, 3.8, 'Integração\n(soma ponderada)', fontsize=9, ha='center')

# Comparação com limiar
ax1.plot([2.4, 3.5], [5, 5], 'k-', linewidth=2)
ax1.plot([3.5, 3.5], [4.5, 5.5], 'r--', linewidth=2, label='Limiar')
ax1.text(3.8, 5, 'Limiar?', fontsize=9)

# Saída
ax1.add_patch(Circle((5, 5), 0.3, color='red', ec='black', linewidth=2))
ax1.text(5, 5, '>', fontsize=10, ha='center', va='center', fontweight='bold', color='white')
ax1.scatter(5.8, 5, s=100, marker='>', color='red', zorder=5)
ax1.text(6.5, 5, 'Dispara?\n(0 ou 1)', fontsize=9)

ax1.text(2, 1, 'Processo: Recebe → Integra → Compara → Dispara', 
        fontsize=9, ha='center', bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))

# NEURÔNIO ARTIFICIAL
ax2.set_xlim(-1, 8)
ax2.set_ylim(-1, 8)
ax2.axis('off')
ax2.set_title('Neurônio Artificial', fontsize=12, fontweight='bold', pad=10)

# Entradas
for i in range(3):
    y = 6 - i*1.5
    ax2.plot([0, 1], [y, 5], 'b-', linewidth=2)
    ax2.scatter(0, y, s=80, color='blue', zorder=5)
    ax2.text(-0.7, y, f'x₁={1if i==0 else 0}', fontsize=9)

# Pesos
for i in range(3):
    y = 6 - i*1.5
    mid_x = 0.5
    mid_y = (y + 5) / 2
    ax2.text(mid_x, mid_y+0.2, f'w{i+1}', fontsize=8, style='italic',
            bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.7))

# Soma
ax2.add_patch(Circle((2, 5), 0.4, color='gold', ec='black', linewidth=2))
ax2.text(2, 5, 'Σ', fontsize=12, ha='center', va='center', fontweight='bold')
ax2.text(2, 3.8, 'z = Σ(w·x) + b', fontsize=9, ha='center',
        bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))

# Ativação
ax2.plot([2.4, 3.5], [5, 5], 'k-', linewidth=2)
ax2.add_patch(FancyBboxPatch((3.5, 4.7), 0.8, 0.6, 
                             boxstyle='round,pad=0.1',
                             edgecolor='black', facecolor='lightgreen', linewidth=2))
ax2.text(3.9, 5, 'σ(z)', fontsize=9, ha='center', va='center', fontweight='bold')

# Saída
ax2.scatter(5.5, 5, s=150, color='red', marker='o', edgecolor='black', linewidth=2, zorder=5)
ax2.text(5.5, 5, 'ŷ', fontsize=10, ha='center', va='center', fontweight='bold')
ax2.text(6.5, 5, '0 a 1\n(contínuo)', fontsize=9)

ax2.text(2, 1, 'Processo: Multiplica → Soma → Ativa → Output', 
        fontsize=9, ha='center', bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.8))

plt.suptitle('Inspiração: O Neurônio Biológico Inspirou o Artificial', 
            fontsize=13, fontweight='bold', y=0.98)
plt.tight_layout()
plt.savefig('results/plots/bio_vs_artificial_neuron.png', dpi=150, bbox_inches='tight')
plt.show()

print("\n✅ Diagrama criado: bio_vs_artificial_neuron.png")
```

## 5. Por que o Paralelo Funciona?

```python
print("\n" + "="*70)
print("✨ POR QUE O PARALELO FUNCIONA?")
print("="*70)

print("""
1. MECANISMO FUNDAMENTAL É IGUAL:
   ✓ Bio: V = Σ(força×sinal) → dispara se > limiar
   ✓ IA:  z = Σ(w×x) + b → ŷ = σ(z)
   → Ambos integram sinais ponderados

2. APRENDIZADO É O MESMO:
   ✓ Bio: plasticidade sináptica (LTP/LTD) ajusta força
   ✓ IA:  backpropagation ajusta pesos
   → Ambos ajustam a "força" da conexão com experiência

3. DOIS NÍVEIS DE PARALELISMO:
   ✓ Bio: 86B neurônios processam em paralelo
   ✓ IA:  bilhões de operações por segundo em GPU
   → Ambos são MASSIVAMENTE paralelos

4. NÃO-LINEARIDADE É ESSENCIAL:
   ✓ Bio: potencial de ação (tudo-ou-nada)
   ✓ IA:  função de ativação (ReLU, Sigmoid, etc)
   → Ambos precisam de não-linearidade para complexidade

5. PLASTICIDADE = APRENDIZADO:
   ✓ Bio: sinapses mudam força com uso (Hebb's rule)
   ✓ IA:  pesos mudam gradualmente (gradient descent)
   → Ambos aprendem ajustando conexões
""")

print("\n" + "="*70)
print("⚠️ MAS TEM LIMITAÇÕES")
print("="*70)

print("""
1. VELOCIDADE:
   ✗ Bio: 200 Hz (muito lento!)
   ✓ IA:  1 Petaflop/s (1.000.000x mais rápido)
   → RNA artificial é MUITO mais rápida para cálculos

2. ROBUSTEZ:
   ✓ Bio: tolerant a ruído, falhas, mudanças
   ✗ IA:  frágil, precisa dados perfeitos
   → Cérebro é MUITO mais robusto

3. GENERALIZAÇÃO:
   ✓ Bio: aprende com poucos exemplos
   ✗ IA:  precisa de milhões de exemplos
   → Cérebro aprende mais eficientemente

4. ENERGIA:
   ✓ Bio: ~20W (tudo no corpo)
   ✗ IA:  kilowatts (GPUs consomem muita energia)
   → Cérebro é MUITO mais eficiente

5. INTERPRETABILIDADE:
   ✓ Bio: podemos estudar bioquímica
   ✗ IA:  "caixa preta" (por isso Grad-CAM!)
   → Ainda não entendemos por que RNA funciona
""")

print("\n" + "="*70)
```
